```python
# self – Instance Reference

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):
        print(f"Name: {self.name}, Marks: {self.marks}")
```

```python
# cls – Class Reference

class Counter:
    count = 0
    def __init__(self):
        Counter.count += 1
    @classmethod
    def show_count(cls):
        print(f"Total objects: {cls.count}")
```

```python
# Public Variables and Methods

class Car:
    def __init__(self, brand):
        self.brand = brand  # this is public variable
    def start(self):  # This is public method
        print(f"{self.brand} car started")
```

```python
# Class Variables & Class Methods

class Bank:
    bank_name = "Faysal Bank"
    @classmethod
    def change_bank_name(cls, name):
        cls.bank_name = name
```

```python
# Static Methods

class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b
```

```python
# Constructors and Destructors

class Logger:
    def __init__(self):
        print("Logger started")
    def __del__(self):
        print("Logger stopped")
```

```python
# Access Modifiers

class Employee:
    def __init__(self, name, salary, ssn):
        self.name = name
        self._salary = salary
        self.__ssn = ssn

emp = Employee("Hamza", 50000, "123")
print(emp.name)       # this is public
print(emp._salary)    # this is protected
# print(emp.__ssn)    # this will give error because it's private
```

```python
# super() Function

class Person:
    def __init__(self, name):
        self.name = name

class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject
```

```python
# Abstract Classes

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, l, w):
        self.l = l
        self.w = w

    def area(self):
        return self.l * self.w
```

```python
# Instance Methods

class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} says woof")
```

```python
# Class Methods

class Book:
    total_books = 0

    @classmethod
    def increment_book_count(cls):
        cls.total_books += 1
```

```python
# Static Methods

class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(c):
        return (c * 9/5) + 32
```

```python
# Composition

class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self, engine):
        self.engine = engine

    def start(self):
        self.engine.start()
```

```python
# Aggregation

class Employee:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, employee):
        self.employee = employee
```

```python
# MRO and Diamond Inheritance

class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(A):
    def show(self):
        print("C")

class D(B, C):
    pass

d = D()
d.show()
```

```python
# Function Decorators

def log_function_call(func):
    def wrapper():
        print("Function is being called")
        func()
    return wrapper

@log_function_call
def say_hello():
    print("hello")
```

```python
# Class Decorators

def add_greeting(cls):
    def greet(self):
        return "Hello Decorator"
    cls.greet = greet
    return cls

@add_greeting
class Person:
    pass
```

```python
# Property Decorators

class Product:
    def __init__(self, price):
        self._price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        self._price = value

    @price.deleter
    def price(self):
        del self._price
```

```python
# callable() and __call__()


class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, x):
        return x * self.factor

m = Multiplier(3)
print(callable(m))
print(m(5))
```

```python
# Custom Exception

class InvalidAgeError(Exception):
    pass

def check_age(age):
    if age < 18:
        raise InvalidAgeError("Age must be 18 or older")

try:
    check_age(16)
except InvalidAgeError as e:
    print(e)
```

```python
# Custom Iterable

class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val
```
